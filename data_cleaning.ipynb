```python
import pandas as pd

# 1. قراءة البيانات
df = pd.read_csv("bank_churn.csv")

# 2. استبعاد الأعمدة غير الضرورية للتحليل (مثل رقم السجل واسم العائلة)
cols_to_drop = [col for col in ['RowNumber', 'CustomerId', 'Surname'] if col in df.columns]
df = df.drop(columns=cols_to_drop)

# 3. إنشاء أعمدة جديدة (Feature Engineering)
# تقسيم الأعمدة إلى فئات عمرية
df['AgeGroup'] = pd.cut(df['Age'], bins=[17, 30, 40, 50, 60, 100], labels=['18-30', '31-40', '41-50', '51-60', '60+'])

# إنشاء عمود يوضح هل العميل لديه رصيد أم رصيده صفر
df['HasBalance'] = df['Balance'].apply(lambda x: 'Balance > 0' if x > 0 else 'Zero Balance')

# 4. حفظ البيانات المنظفة في ملف جديد
df.to_csv("cleaned_bank_churn.csv", index=False)

print("Data cleaned successfully! Shape:", df.shape)
df.head()
```
